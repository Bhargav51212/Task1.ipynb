```python
# Take username input from the user
username = input("Enter username: ")
# Take password input from the user
password = input("Enter Password: ")
# Check if both username and password match the predefined values
if username == "admin" and password == "1234":
    # This block executes if credentials are correct
    print("Login Successful")
else:
    # This block executes if credentials are incorrect
    print("Invalid Credentials")
```

```python
# List of student marks
marks = [45, 78, 90, 33, 60]
# Count for passed and failed students
passed = 0
failed = 0
for i in marks:
    # Check if student passed (marks >= 50)
    if i >= 50:
        passed += 1   # Increment pass count
    else:
        failed += 1   # Increment fail count
# Print total number of students
print("Total Students:", len(marks))
# Print number of passed students
print("Number of Pass Students:", passed)
# Print number of failed students
print("Number of Fail Students:", failed)
# Check final result
if failed == 0:   # Corrected variable name
    print("Final Result: All students passed")
else:
    print("Final Result: Some students failed.")
```

```python
# List containing names with extra spaces and mixed case letters
raw_names = [" Alice ", "bob", " CHARLIE "]
# Create an empty list to store formatted names
cleaned_names = []
# Process each name in the list
for item in raw_names:
    # Remove spaces and convert to lowercase
    updated_name = item.strip().lower()
    # Add the cleaned name to the new list
    cleaned_names.append(updated_name)
# Display the original and processed lists
print("Names Before Cleaning:", raw_names)
print("Names After Cleaning:", cleaned_names)
```

```python
# List containing  text message
messages= ["Hi", "Welcome to the platform", "OK"]
# Loop through each text in the list
for msg in messages:  
    # Calculate the number of characters in the current text
    char_count = len(msg)   
    # Display the message and its size
    print("Current Message:", msg)
    print("Character Count:", char_count)  
    # Check whether the message is long or short
    if char_count >= 11:
        print("Category: Message exceeds 10 characters")
    else:
        print("Category: Message is within limit")

```

```python
# List representing log entries
logs = ["INFO", "ERROR", "WARNING", "ERROR"]

# Initialize a error counter
error_total = 0
for entry in logs: 
    # Check if the current entry indicates an error
    if entry == "ERROR":
         # Increase error counter
        error_total += 1  
# Display the total number of error messages found
print("Total ERROR entries detected:", error_total)
```
