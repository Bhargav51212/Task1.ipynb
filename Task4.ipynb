```python
# Function to check whether the recharge request is valid
def valid_recharge(amount):
    available_plans = [199, 299, 399, 599]
# Check minimum recharge condition
    if amount < 50:
        print("Recharge failed: Minimum amount should be ₹50 or more.")
        return False
# Check if plan exists
    if amount not in available_plans:
        print("Recharge failed: Selected plan is not available.")
        print("Please choose from:", available_plans)
        return False
# If all conditions are satisfied
    print("Recharge of ₹", amount, "has been processed successfully.")
    return True
# Allow user to retry until correct input is given
while True:
    try:
        recharge_amount = int(input("Enter the recharge amount: ₹"))
        result = valid_recharge(recharge_amount)
        if result:
            break 
        else:
            print("Try again.\n")
    except ValueError:
        print("Invalid input. Please enter numbers only.\n")
```

```python
# Function to check stock levels
def check_inventory(products):
    for item, quantity in products.items():
        if quantity < 15:
            print(item, "→ Reorder Alert (Stock:", quantity, ")")
        else:
            print(item, "→ Stock OK (Stock:", quantity, ")")
# Sample dictionary of products and their stock quantities
inventory_data = {
    "Rice Bags": 20,
    "Sugar Packs": 10,
    "Oil Bottles": 25,
    "Soap Boxes": 12
}
check_inventory(inventory_data)
```

```python
# Function to calculate average and return result
def evaluate_student(marks_list):
    total_marks = 0   
# Calculate total using loop
    for mark in marks_list:
        total_marks += mark
    average = total_marks / len(marks_list)
    if average >= 50:
        return average, "Pass"
    else:
        return average, "Fail"
student_marks = [60, 85, 70, 45]
avg, status = evaluate_student(student_marks)
print("Average Marks:", avg)
print("Result:", status)
```

```python
# Function to calculate cab fare
def calculate_fare(distance, peak_time):
    base_fare = 50
    rate_per_km = 12
    total = base_fare + (distance * rate_per_km)
# Apply peak hour charges
    if peak_time.lower() == "yes":
        total = total + (0.25 * total)
    return total
# Retry system using while loop
while True:
    try:
        km = float(input("Enter distance travelled (in km): "))
        peak = input("Is it peak hour? (yes/no): ")
        if km <= 0:
            print("Distance must be greater than 0. Try again.\n")
            continue
        final_amount = calculate_fare(km, peak)
        print("Total Fare: ₹", round(final_amount, 2))
        break 
    except ValueError:
        print("Invalid input. Please enter numeric value for distance.\n")
```

```python
# Function to calculate attendance percentage
def check_attendance(attendance_list):
    present_count = 0
    # Count number of present days using loop
    for status in attendance_list:
        if status.upper() == "P":
            present_count += 1
    total_days = len(attendance_list)
    percentage = (present_count / total_days) * 100
    if percentage >= 75:
        return percentage, "Eligible"
    else:
        return percentage, "Not Eligible"
records = ["P", "A", "A", "P", "A", "P", "P", "P"]
attendance_percent, result = check_attendance(records)
print("Attendance Percentage:", round(attendance_percent, 2), "%")
print("Status:", result)
```

```python
# Function to validate password strength
def validate_password(password):
    has_digit = False
    has_special = False
    special_chars = "@#$"
# Check minimum length
    if len(password) < 8:
        return "Weak Password: Minimum length should be 8 characters."
# Loop through each character
    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch in special_chars:
            has_special = True
# Final validation
    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password: Must include at least one digit and one special character (@#$)."
user_password = input("Enter your password: ")
result = validate_password(user_password)
print(result)
```
