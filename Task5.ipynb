```python
def analyze(likes_list):
    likes = 0  
    for likes in likes_list:
        likes += likes 
    if likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"  
    print("Total Likes:", likes)
    print("Post Status:", status)
likes_data = [200, 300, 150, 400]  
analyze(likes_data)
```

```python
def medicine_stock(stock):
    print("Medicine Stock:", stock)  
    if stock < 10:
        print("Status: Low Stock Alert")
    else:
        print("Status: Stock Sufficient")
medicine_stock(6)
```

```python
def analyze_rainfall(rainfall_data, required_level):
    total = 0
    # Loop to calculate total rainfall
    for rain in rainfall_data:
        total += rain
    average = total / len(rainfall_data) 
    print("Average Rainfall:", int(average))
    if average >= required_level:
        print("Rainfall Status: Adequate Rainfall")
    else:
        print("Rainfall Status: Inadequate Rainfall")
rainfall = [70, 75, 68, 80, 67]
required_level = 70
analyze_rainfall(rainfall, required_level)
```

```python
def detect_duplicate_usernames(usernames):
    unique_usernames = set(usernames)
    if len(unique_usernames) < len(usernames):
        print("Duplicate Accounts Found: Yes")
        print("Status: Flag accounts")
    else:
        print("Duplicate Accounts Found: No")
        print("Status: Accounts are unique")
users = ["john123", "alice99", "john123", "mike007"]
detect_duplicate_usernames(users)
```

```python
def check_appointment_eligibility(age):
    print("Patient Age:", age) 
    if age >= 18:
        print("Eligibility Status: Eligible")
    else:
        print("Eligibility Status: Not Eligible")
check_appointment_eligibility(21)
```

```python
def filter_premium_crops(prices):
    premium_crops = []
    for price in prices:
        if price > 2000:
            premium_crops.append(price)
    print("Premium Crops:", premium_crops)
crop_prices = [1500, 2500, 1800, 3200, 1999]
filter_premium_crops(crop_prices)
```

```python
def evaluate_application_health(error_count):
    print("Error Count:", error_count)
    if error_count == 0:
        print("System Status: Healthy")
    elif error_count <= 5:
        print("System Status: Minor Issues")
    else:
        print("System Status: Critical Issues")
evaluate_application_health(7)
```

```python
def validate_transaction(amount):
    daily_limit = 50000
    print("Transaction Amount:", amount)  
    if amount <= daily_limit:
        print("Transaction Status: Approved")
    else:
        print("Transaction Status: Rejected")
validate_transaction(60000)
```

```python
def check_exam_eligibility(attendance_list):
    total_days = len(attendance_list)
    present_days = 0
    for status in attendance_list:
        if status == "P":
            present_days += 1
    percentage = (present_days / total_days) * 100
    print("Attendance Percentage:", percentage)
    if percentage >= 75:
        print("Exam Eligibility: Eligible")
    else:
        print("Exam Eligibility: Not Eligible")
attendance = ["P", "P", "A", "P", "P"]  # 4 out of 5 present = 80%
check_exam_eligibility(attendance)
```

```python
def calculate_electricity_bill(units):
    bill = 0
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)
    if bill < 500:
        usage_status = "Low Usage"
    elif bill <= 1500:
        usage_status = "Moderate Usage"
    else:
        usage_status = "High Usage"
    return bill, usage_status
units_consumed = 250
total_bill, status = calculate_electricity_bill(units_consumed)
print("Units Consumed:", units_consumed)
print("Total Bill: ₹", total_bill)
print("Usage Classification:", status)
```
