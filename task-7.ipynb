```python
class smartlight:
    def __init__(self,name,status):
        self.name = name
        self.status = status
    def show_status(self):
        print(self.name, "is", self.status)
light = smartlight("Bedroom Light", "Off")
light.show_status()    
```

```python
class EmployeeID:
    def __init__(self,name,emp_id,department):
        self.name = name
        self.emp_id = emp_id
        self.department = department
    def show_result(self):
        print("Employee ID Card")
        print("Name:",self.name)
        print("ID:",self.emp_id)
        print("Department:",self.department)
employee = EmployeeID("Bhargav","EMP101","IT")
employee.show_result()
```

```python
class MobileContact:
    def __init__(self,name,phn_no):
        self.name = name
        self.phn_no = phn_no
    def show_result(self):
        print("Contact Saved")
        print("Name:",self.name)
        print("Phone Number:",self.phn_no)
contact_info = MobileContact("Bhargav", 1234567890)
contact_info.show_result()
```

```python
class Product:
    def __init__(self,name,price):
        self.name = name
        self.price = price
    def show_result(self):
        print("Product:",self.name)
        print("Price:",self.price)
product_info = Product("Laptop",4999)
product_info.show_result()
```

```python
class MovieRating:
    def __init__(self,name,rating):
        self.name = name
        self.rating = rating
    def show_result(self):
        print("Movie Name:",self.name)
        print("Rating:",self.rating,"/5")
result = MovieRating("Raja Saab",4)
result.show_result()
```

```python
class DeliveryAddress:
    def __init__(self,customer_name,Address):
        self.customer_name = customer_name
        self.Address = Address
    def result(self):
        print("Delivery Details")
        print("Customer:",self.customer_name)
        print("Address:",self.Address)
results = DeliveryAddress("Riya","Mumbai")
results.result()
```
